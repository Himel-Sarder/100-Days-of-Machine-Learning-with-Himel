## To perform polynomial regression on the given data and find the coefficients using the formula a = X^-1.B

# Build-in function for perform polynomial regression

```python
def calculate_coefficients(X_values, y_values):
    """
    Calculate the coefficients of a second-degree polynomial (y = a0 + a1*x + a2*x^2)
    using the least squares method with matrix operations.
    
    Parameters:
    - X_values: List of x values (independent variable).
    - y_values: List of y values (dependent variable).
    
    Returns:
    - coefficients: List of coefficients [a0, a1, a2].
    """
    # Calculate necessary sums
    n = len(X_values)
    
    sum_x = sum(X_values)
    sum_x2 = sum(x ** 2 for x in X_values)
    sum_x3 = sum(x ** 3 for x in X_values)
    sum_x4 = sum(x ** 4 for x in X_values)
    
    sum_y = sum(y_values)
    print(sum_y)
    
    sum_xy = sum(x * y for x, y in zip(X_values, y_values))
    print(sum_xy)
    
    sum_x2y = sum(x**2 * y for x, y in zip(X_values, y_values))
    print(sum_x2y)
    
    # Construct matrix X and vector B
    X_matrix = np.array([[n, sum_x, sum_x2],
                         [sum_x, sum_x2, sum_x3],
                         [sum_x2, sum_x3, sum_x4]])

    print('X Matrix: \n', X_matrix)
    
    B_vector = np.array([sum_y, sum_xy, sum_x2y])
    print('\nB :', B_vector)

    X_inv = np.linalg.inv(X_matrix)
    print('Inverse of X : ', X_inv)
    
    # Calculate the coefficients using X^-1 * B
    coefficients = X_inv.dot(B_vector)
    
    return coefficients
```

```python
# Compute the coefficients
coefficients = calculate_coefficients(X_values, y_values)

# Display the coefficients
print("\nCoefficients (a0, a1, a2, ...):")
print(coefficients)
```

# Another way to create a polynomial Regression function for changing Degree

```python
def polynomial_regression(X, y, degree):
    """
    POLYNOMIAL_REGRESSION: Computes the coefficients for polynomial regression.
    Input:
        X: Column vector or array-like of independent variable (e.g., Level)
        y: Column vector or array-like of dependent variable (e.g., Salary)
        degree: Degree of the polynomial (e.g., 2 for quadratic)
    Output:
        coefficients: Coefficients of the polynomial [a0, a1, a2, ..., an]
    """
    # Ensure X and y are numpy arrays
    X = np.array(X)
    y = np.array(y)

    # Check if X and y have the same length
    if len(X) != len(y):
        raise ValueError("X and y must have the same length.")

    # Construct the design matrix X_poly
    X_poly = np.column_stack([X**i for i in range(degree + 1)])

    # Compute the coefficients using the least squares method
    coefficients, _, _, _ = np.linalg.lstsq(X_poly, y, rcond=None)

    return coefficients
```

```python
# Degree of the polynomial (e.g., 2 for quadratic)
degree = 4

# Compute the coefficients
coefficients = polynomial_regression(X_values, y_values, degree)

# Display the coefficients
print("Coefficients (a0, a1, a2, ...):")
print(coefficients)
```
