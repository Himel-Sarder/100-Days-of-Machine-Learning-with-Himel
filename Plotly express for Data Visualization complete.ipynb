```python
import plotly.express as px
```

```python
df = pd.read_csv('/kaggle/input/bangladesh-districts-wise-population/city_population.csv')
df.sample(10)
```

```python
# Simple Scatter Plot

fig = px.scatter(
    df,
    x='Established',
    y='Area (km2)',
    color='Name',
    title='Established vs Area (km2) Scatter Plot',
    labels={'Established': 'Estalished', 'Area (km2)': 'Area (km2)'}
)

fig.show()
```

```python
# Add Population_1991 column with size parameter

fig = px.scatter(
    df,
    x='Established',
    y='Area (km2)',
    color='Name',
    size='Population_1991',
    title='Established vs Area (km2) Scatter Plot',
    labels={'Established': 'Estalished', 'Area (km2)': 'Area (km2)'}
)

fig.show()
```

```python
# Increase size with size_max Parameter

fig = px.scatter(
    df,
    x='Area (km2)',
    y='Established',
    color='Name',
    size='Population_1991',
    size_max=50,
    title='Established vs Area (km2) Scatter Plot',
    labels={'Established': 'Estalished', 'Area (km2)': 'Area (km2)'}
)

fig.show()
```

```python
# Increase axis length with range_y and range_x

fig = px.scatter(
    df,
    x='Area (km2)',
    y='Established',
    color='Name',
    size='Population_1991',
    size_max=50,
    range_x=[0, 7000],      
    range_y=[1600, 2025],
    opacity=0.6,
    title='Established vs Area (km2) Scatter Plot',
    labels={'Established': 'Estalished', 'Area (km2)': 'Area (km2)'}
)

fig.show()
```

```python
# Different Shape of differenct division

fig = px.scatter(
    df,
    x='Area (km2)',
    y='Established',
    color='Division',
    size='Population_1991',
    size_max=50,
    range_x=[0, 7000],      
    range_y=[1600, 2025],
    symbol='Division',
    opacity=0.6,
    title='Established vs Area (km2) Scatter Plot',
    labels={'Established': 'Estalished', 'Area (km2)': 'Area (km2)'}
)

fig.show()
```

```python
# Reshape population columns into long format

df_long = df.melt(
    id_vars=['Name', 'Abbr.', 'Division', 'Established', 'Native', 'Area (km2)'],
    value_vars=['Population_1991', 'Population_2001', 'Population_2011', 'Population_2022'],
    var_name='year',
    value_name='population'
)

# Clean the 'year' column by extracting only the numeric year
df_long['year'] = df_long['year'].str.extract('(\d+)$').astype(int)

df_long.head()
```

```python
fig = px.scatter(
    df_long,
    x='Area (km2)',
    y='population',
    color='Division',          
    size_max=50,
    opacity=0.7,
    animation_frame='year',
    animation_group='Name',      
    range_y=[0, 16000000],
    title='Population Over Time by Division',
    labels={
        'year': 'Year',
        'population': 'Population',
        'Division': 'Division'
    }
)

fig.show()
```

```python
# Filter for Dhaka Division
temp = df_long[df_long['Division'] == 'Dhaka']

# Plot line chart of population over year

fig = px.line(
    temp,
    x='year',
    y='population',
    color='Name', 
    title="Population Growth Over Time – Dhaka Division"
)

fig.show()
```

```python
# Bar Plot of contribution of each division on total population over year

fig = px.bar(df_long, x="year", y="population", color="Division")
fig.show()
```

```python
# Bar Plot of each division over years

fig = px.bar(df_long, x="year", y="population", color="Division", barmode="group")
fig.show()
```

```python
# Group population per Division per Year
df_grouped = df_long.groupby(['year', 'Division'], as_index=False)['population'].sum()

fig = px.bar(
    df_grouped,
    x='year',
    y='population',
    color='Division',
    barmode='group',    
    text_auto=True,
    title='Total Population per Division Over Years'
)

fig.show()
```

```python
# Group population per Division per Year
df_grouped = df_long.groupby(['year', 'Division'], as_index=False)['population'].sum()

fig = px.bar(
    df_grouped,
    x='year',
    y='population',
    color='Division',
    barmode='group',    
    text_auto=True,
    title='Total Population per Division Over Years'
)

fig.show()
```

```python
# Population Growth by District (Grouped by Division

fig = px.bar(
    df_long,
    x='Name',                  # Districts
    y='population',            # Population value
    color='Division',          # Color by division
    animation_frame='year',    # Animate over year
    animation_group='Name',    # Track bars by district
    text='population',         # Show population value on bars
    range_y = [0, 16000000],
    title='Population Growth by District (Grouped by Division)',
    labels={'Name': 'District', 'population': 'Population'},
    height=600
)

# Improve layout
fig.update_layout(
    xaxis_tickangle=45,
    uniformtext_minsize=8,
    uniformtext_mode='hide'
)

fig.show()
```

```python
# Distribution of Districts by Year of Establishment

fig = px.histogram(
    df,
    x='Established',
    nbins=30,  # Adjust number of bins for grouping
    title='Distribution of Districts by Year of Establishment',
    labels={'Established': 'Year Established'},
    text_auto=True,
    color='Division'
)

fig.show()
```

```python
import plotly.graph_objects as go

# Extract labels and values
labels = temp_2022_Dhaka['Name']
values = temp_2022_Dhaka['population']

# Create pull list: 0.1 for Gazipur, 0 for others
pull = [0.1 if name == 'Gazipur' else 0 for name in labels]

# Create pie chart using go
fig = go.Figure(data=[
    go.Pie(
        labels=labels,
        values=values,
        pull=pull,
        hoverinfo='label+percent+value',
        textinfo='label+percent',
        textposition='inside'
    )
])

fig.update_layout(
    title='Population Contribution of Districts in Dhaka Division (2022) - Gazipur Highlighted'
)

fig.show()
```
