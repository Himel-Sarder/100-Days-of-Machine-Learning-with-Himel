# Outliers using Winsorization Technique

Winsorization is a **data transformation technique** used to limit extreme values (outliers) in a dataset by **replacing them with less extreme values** instead of removing them. This helps reduce the effect of outliers while preserving the overall data distribution.  

### **How Winsorization Works**  
1. **Set the Percentile Thresholds**  
   - Choose a percentage of extreme values to be modified.  
   - Common choices: **5% Winsorization (95% Winsorization level)** means the **top 5%** and **bottom 5%** of values are replaced.  

2. **Modify Extreme Values**  
   - Replace values **above the upper percentile** with the value at that percentile.  
   - Replace values **below the lower percentile** with the value at that percentile.  


### **Example**  
**Dataset:**  
2, 5, 7, 10, 15, 20, 50, 100, 200

**Winsorizing at 10%:**  
- Bottom 10% (lowest value 2) → Replaced by next percentile value (5).  
- Top 10% (highest value 200) → Replaced by next lower value (100).  

**Modified Dataset:**  
5, 5, 7, 10, 15, 20, 50, 100, 100


### **Why Use Winsorization?**  
✅ **Reduces the impact of extreme outliers** while keeping all data points.  
✅ **Useful for skewed data** (e.g., financial data, income distribution).  
✅ **Maintains statistical properties** better than removing outliers.  


### **Types of Winsorization**  
1. **Symmetric Winsorization** – Both ends of the distribution are adjusted.  
2. **Asymmetric Winsorization** – Only one end (upper or lower) is adjusted.  


### **Difference from Trimming**  
- **Trimming:** Completely removes extreme values.  
- **Winsorization:** Replaces extreme values with threshold values.  

This makes Winsorization useful when **removing data is not an option** but reducing the effect of outliers is necessary.

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('weight-height.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df['Height'].describe()
```

```python
import seaborn as sns
```

```python
sns.distplot(df['Height'])
```

```python
sns.boxplot(df['Height'])
```

```python
upper_limit = df['Height'].quantile(0.99)
upper_limit
```

```python
lower_limit = df['Height'].quantile(0.01)
lower_limit
```

```python
new_df = df[(df['Height'] <= 74.78) & (df['Height'] >= 58.13)]
```

```python
new_df['Height'].describe()
```

```python
df['Height'].describe()
```

```python
sns.distplot(new_df['Height'])
```

```python
sns.boxplot(new_df['Height'])
```

```python
# Capping --> Winsorization
df['Height'] = np.where(df['Height'] >= upper_limit,
        upper_limit,
        np.where(df['Height'] <= lower_limit,
        lower_limit,
        df['Height']))
```
