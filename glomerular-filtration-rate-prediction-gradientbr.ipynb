# **Without Using Learning Rate**

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np
import matplotlib.pyplot as plt

# Initialize with mean prediction
df = X.copy()
df['y_true'] = y
initial_prediction = y.mean()
print("1st Model's Prediction(Mean) : ")
print(y.mean())
df['total_pred'] = initial_prediction
residual = y - df['total_pred']   # residuals after mean

n_iter = 40  # number of boosting steps
trees = []   # to store fitted trees
results = [] # to store metrics per iteration

for i in range(n_iter):
    # Fit tree on current residuals
    tree = DecisionTreeRegressor(max_leaf_nodes=5, random_state=42)
    tree.fit(X, residual)
    
    # Predict
    pred = tree.predict(X)
    
    # Save predictions
    df[f'pred{i+1}'] = pred
    
    # Update total prediction
    df['total_pred'] += pred
    
    # Update residual
    residual = y - df['total_pred']
    df[f'res{i+1}'] = residual

    
    # Save tree
    trees.append(tree)
    
    # Evaluate after this iteration
    y_pred = df['total_pred']
    r2 = r2_score(y, y_pred)
    mse = mean_squared_error(y, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y, y_pred)
    
    results.append({
        "Model": f"Tree {i+1}",
        "R²": r2,
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae
    })

# Print results for each tree
print("Performance after each tree")
print("*" * 60)
for r in results:
    print(f"{r['Model']}: R²={r['R²']:.4f}, MSE={r['MSE']:.4f}, RMSE={r['RMSE']:.4f}, MAE={r['MAE']:.4f}")

# Show all trees
for i, tree in enumerate(trees, start=1):
    plt.figure(figsize=(12,6))
    plot_tree(tree, filled=True, fontsize=8)
    plt.title(f"Decision Tree {i}")
    plt.show()
```

# **With Learning Rate**

```python
# Initialize with mean prediction
initial_prediction = y.mean()
df['cumulative_pred'] = initial_prediction

# Set learning rate and number of trees
learning_rate = 0.1
n_trees = 250

for i in range(n_trees):
    # Calculate residuals
    residuals = y - df['cumulative_pred']
    
    # Fit tree to residuals
    tree = DecisionTreeRegressor(max_leaf_nodes=5, max_depth=3)
    tree.fit(X, residuals)
    
    # Update predictions with learning rate
    df['cumulative_pred'] += learning_rate * tree.predict(X)
    
    # Calculate metrics periodically
    if (i+1) % 10 == 0:
        r2 = r2_score(y, df['cumulative_pred'])
        print(f"After {i+1} trees, R²: {r2:.4f}")

# Final evaluation
final_r2 = r2_score(y, df['cumulative_pred'])
final_mse = mean_squared_error(y, df['cumulative_pred'])
print(f"\nFinal R²: {final_r2:.4f}")
print(f"Final MSE: {final_mse:.4f}")

```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Initialize with mean prediction
df['cumulative_pred'] = y.mean()  # This is your pred1
current_prediction = df['cumulative_pred'].copy()

# Set parameters
n_trees = 250
learning_rate = 0.1
max_leaf_nodes = 5

# Store metrics for plotting
r2_scores = []
mse_scores = []

for i in range(n_trees):
    # Calculate residuals (actual - current prediction)
    residuals = y - current_prediction
    
    # Fit tree to residuals
    tree = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=42)
    tree.fit(X, residuals)
    
    # Get tree predictions
    tree_pred = tree.predict(X)
    
    # Update cumulative prediction with learning rate
    current_prediction += learning_rate * tree_pred
    
    # Store metrics every 10 iterations
    if (i + 1) % 10 == 0 or i == 0:
        r2 = r2_score(y, current_prediction)
        mse = mean_squared_error(y, current_prediction)
        r2_scores.append(r2)
        mse_scores.append(mse)
        print(f"Iteration {i+1}: R² = {r2:.4f}, MSE = {mse:.4f}")

# Final evaluation
final_r2 = r2_score(y, current_prediction)
final_mse = mean_squared_error(y, current_prediction)
final_rmse = np.sqrt(final_mse)
final_mae = mean_absolute_error(y, current_prediction)

print("\n" + "="*50)
print("FINAL RESULTS:")
print("="*50)
print(f"R² Score: {final_r2:.4f}")
print(f"MSE: {final_mse:.4f}")
print(f"RMSE: {final_rmse:.4f}")
print(f"MAE: {final_mae:.4f}")

# Plot convergence
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(range(0, n_trees + 1, 10), r2_scores, 'b-o', linewidth=2)
plt.xlabel('Iteration')
plt.ylabel('R² Score')
plt.title('R² Score Convergence')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(range(0, n_trees + 1, 10), mse_scores, 'r-o', linewidth=2)
plt.xlabel('Iteration')
plt.ylabel('MSE')
plt.title('MSE Convergence')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Plot the last tree (for visualization)
plt.figure(figsize=(20, 10))
plot_tree(tree, filled=True, feature_names=X.columns, rounded=True)
plt.title(f'Tree at iteration {n_trees}')
plt.show()

# Show final predictions vs actual
plt.figure(figsize=(10, 6))
plt.scatter(y, current_prediction, alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', linewidth=2)
plt.xlabel('Actual GFR')
plt.ylabel('Predicted GFR')
plt.title('Actual vs Predicted GFR (Final Model)')
plt.grid(True, alpha=0.3)
plt.show()
```
