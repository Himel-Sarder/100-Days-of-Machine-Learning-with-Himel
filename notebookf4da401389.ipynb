```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('/kaggle/input/multiple-linear-regression-dataset/multiple_linear_regression_dataset.csv')
df.head()
```

```python
X = df[['age', 'experience']].values
y = df['income'].values
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Create scalers
X_scaler = StandardScaler()
y_scaler = StandardScaler()

# Scale X and y
X_scaled = X_scaler.fit_transform(X_train)
y_scaled = y_scaler.fit_transform(y_train.reshape(-1, 1)).flatten()  # Make y 1D again
```

```python
reg = LinearRegression()
reg.fit(X_scaled,y_scaled)
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
class GDRegressor:
    
    def __init__(self, learning_rate=0.01, epochs=1000):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.losses = []

    def fit(self, X_train, y_train):
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            y_hat = np.dot(X_train, self.coef_) + self.intercept_
            intercept_der = -2 * np.mean(y_train - y_hat)
            coef_der = -2 * np.dot((y_train - y_hat), X_train) / X_train.shape[0]
            
            self.intercept_ -= self.lr * intercept_der
            self.coef_ -= self.lr * coef_der
            
            loss = np.mean((y_train - y_hat) ** 2)
            self.losses.append(loss)

        print("Final intercept:", self.intercept_)
        print("Final coefficients:", self.coef_)

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python


# Train model
model = GDRegressor(learning_rate=0.01, epochs=1000)
model.fit(X_scaled, y)
```

```python
# Predict
y_pred = model.predict(X_scaled)

# Evaluate
print("Predictions:", y_pred)
print("R² Score:", r2_score(y, y_pred))
```
