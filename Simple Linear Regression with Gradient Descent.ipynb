```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/kaggle/input/salary-dataset-simple-linear-regression/Salary_dataset.csv')
```

```python
df.head()
```

```python
X = df[['YearsExperience']]
```

```python
y = df[['Salary']]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split

train_X,test_X,train_y,test_y=train_test_split(X,y,test_size=0.2,shuffle=True)
```

```python
train_X.shape
```

```python
import matplotlib.pyplot as plt

plt.scatter(train_X,train_y)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(train_X,train_y)
```

```python
predicted_values= model.predict(X)
plt.scatter(X,y)
plt.plot(X,predicted_values,color="red")
```

```python
from sklearn.metrics import r2_score

y_predicted = model.predict(test_X)
r2=r2_score(test_y, y_predicted)
print(r2)
```

```python
model.fit(train_X, train_y)
print(model.coef_)
print(model.intercept_)
```

```python
# Lets apply Gradient Descent assuming slope is constant m = 9319.94022518
# and let's assume the starting value for intercept b = 0
X_array = X.to_numpy().flatten()
y_pred = 9319.94022518 * X_array + 100
```

```python
plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='#00a65a',label='b = 0')
plt.legend()
plt.show()
```

```python
m = 9319.94022518
b = 100

loss_slope = -2 * np.sum(y.values.ravel() - m * X.values.ravel() - b)

loss_slope
```

```python
# Lets take learning rate = 0.01
lr = 0.01

step_size = loss_slope*lr
step_size
```

```python
# Calculating the new intercept
b = b - step_size
b
```

```python
y_pred1 = (9319.94022518 * X.values.ravel()) + b

plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='#00a65a',label='b (pred1)')
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
```

```python
# Initial values
m = 9319.94022518  # fixed slope
b = 0              # initial intercept
lr = 0.01          # learning rate
epochs = 100       # number of iterations

# For tracking b values over epochs (optional)
b_values = []

for i in range(epochs):
    # Compute the gradient (loss_slope)
    loss_slope = -2 * np.sum(y.values.ravel() - m * X.values.ravel() - b)

    # Step size
    step_size = loss_slope * lr

    # Update intercept
    b = b - step_size

    # Store for analysis
    b_values.append(b)

    # Optional: Print intermediate values
    print(f"Epoch {i+1:3}: b = {b:.4f}, step size = {step_size:.4f}, loss_slope = {loss_slope:.4f}")
```
